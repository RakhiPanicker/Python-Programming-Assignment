1. Create a 1D NumPy Array:

```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

Inspection and Properties
1. Shapes

```python
print(temperatures_w1.shape)
```

2. Data Type (dtype)


```python
print(temperatures_w1.dtype)
```

3. Number of elements in the Array

```python
print(temperatures_w1.size)
```

3. Array Operations:

```python
temperatures_f = (temperatures_w1 * 9/5) + 32                      # Formula used -- > Fahrenheit = (Celsius × 9/5) + 32

print(temperatures_f)
```

Find the maximum, minimum, and mean temperatures for the week.

```python
print(np.max(temperatures_w1))             # To find maximum temperature
```

```python
print(np.min(temperatures_w1))       # To find Minimum temperature
```

```python
print(np.mean(temperatures_w1))      # To find the average or mean temperature
```

4. Array Slicing and Indexing:

```python
print(temperatures_w1[:3])        # temperatures for the first three days of the week.
```

```python
print(temperatures_w1[-2:])       # temperatures for the last two days.
```

```python
print(temperatures_w1[2:5])           # temperatures for the middle three days of the week.
```

5. Create a 2D Array:

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],             # Temperature of 2 different weeks
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

6. Inspect and Slice the 2D Array:

```python
print(temperatures.shape)     # Shapes
print(temperatures.dtype)     # Datatype
print(temperatures.size)      # Total number of elements in temperature array that is size
```

Extract the temperatures for each week and the weekends (last two days) for both weeks.

```python
print(temperatures[0])              # week 1 temperature
```

```python
print(temperatures[1])              # week 2 temperature
```

```python
print(temperatures[0, -2:])         # Last 2 days of week 1, Temperature(0) represents week 1
```

```python
print(temperatures[1, -2:])         # Last 2 days of week 2, Temperature(1) represents week 2
```

Pandas Series:
1. Creating Pandas Series:

```python
import pandas as pd
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

2. Indexing and Slicing:

```python
print(marks.iloc[0])       # .iloc will tell the integer location and [0] will indicate first position
```

```python
# marks = pd.Series([95, 90, 88, 85, 80], index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"])

marks.loc[["Rank1", "Rank2", "Rank3"]]    # .loc is used when we want to select data using the index LABEL.Here we have to find top 3 Rank
```

```python
marks.iloc[2]      # .loc[2] will give the value which is at position 2. 
```

```python
marks[marks > 90]      # To find the marks greater than 90
```

3. Manipulating Series:

```python
marks.loc["Rank1"] = 100          # to change Rank1 from 95 to 100
```

```python
marks          # This will give the output after updating Rank1 as 100.
```

```python
marks = marks.drop("Rank5")        # remove the entry whose label is Rank5.
```

```python
marks     # output shows only entry till Rank4
```

```python
cgpa = marks / 10     # CGPA means Cumulative Grade Point Average    100/10 , 92/10 , 89/10 , 85/10  --> 10.0, 9.2, 8.9, 8.5
```

```python
cgpa    # CGPA is a grade-point value used to represent a student's academic performance.  
```

Pandas DataFrame:
1. Creating Pandas DataFrame:

```python
import pandas as pd

transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": ["Electronics", "Clothing", "Electronics", "Furniture", "Clothing", "Electronics", "Furniture", "Clothing", "Furniture", "Electronics"],
    "Region": ["North", "South", "North", "East", "West", "North", "East", "West", "South", "North"],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
```

```python
transactions    # This will give the output separately as shown below
```

2. Data Exploration:

```python
transactions.head()        # head() means "show me the beginning of the DataFrame." that is first 5 Rows
```

```python
transactions.tail()    # tail() means this will show the last 5 rows of our DataFrame.
```

```python
transactions.shape        # This tells us how many rows and columns are in our DataFrame.
```

```python
transactions.columns        # This will show the names of all 4 columns.
```

```python
transactions.dtypes   # This will tell us what type of data is stored in each column.
```

```python
transactions[["ProductCategory", "Amount"]]          #  Display only the 'ProductCategory' and 'Amount' columns.
```

```python
transactions.iloc[:, -3:]        # using iloc for this because we're selecting columns by their position.
```

```python
transactions[(transactions["Region"] == "North") & (transactions["Amount"] > 200)]     # Region = North , Amount > 200
```

```python
transactions["ProductCategory"].value_counts()            # to Count how many times each product category appears.
```

```python
transactions["Region"].unique()      # This means showing each region only once, without duplicates.
```

```python
transactions.groupby("Region")["Amount"].mean()        # For each region we have to calculate the average (mean) transaction amount.
```

3. Manipulating the DataFrame:

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165  

#  We need to change the Amount of transaction 102 to 165. Find TransactionID 102 → go to its Amount → change it to 165.
```

```python
transactions[transactions['TransactionID'] == 102]       # previously the amount for 102 row was 150, we changed it to 165
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10          # Add new column "Discount" which shows the discounted amount. 
```

```python
transactions    # To check the output 
```

Remove the row with TransactionID 109.

```python
transactions = transactions[transactions['TransactionID'] != 109]
```

```python
transactions
```

Delete the 'Discount' column from the DataFrame.

```python
transactions = transactions.drop('Discount', axis=1)     # to delete the discount column use "drop".
```

```python
transactions          # check the output, here the Discount column is removed
```
